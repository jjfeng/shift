# Subset masks

```python
def group_to_index(a):
    print(a[["group_idx","group_name"]].drop_duplicates())
    for g in a["group_idx"].unique():
        indices = a[a["group_idx"]==g].index.tolist()
        subgroup_mask = np.ones(a.shape[0]).astype(int)
        for i in indices:
            subgroup_mask[i] = 0
        print(",".join(map(str, subgroup_mask)))

def group_to_individual_feature_index(a, g):
    print(a[a["group_idx"]==g]["var_name"])
    if (g >= 0) and (g < len(a["group_name"].unique())):
        indices = a[a["group_idx"]==g].index.tolist()
    else:
        indices = a.index.tolist()
    print(a.iloc[indices]["group_name"].unique())
    for i in indices:
        subgroup_mask = np.ones(a.shape[0]).astype(int)
        subgroup_mask[i] = 0
        print(",".join(map(str, subgroup_mask)))
```
